## 4. 간단한 토큰 빈도와 어휘 다양성

빈도 분석은 코퍼스의 주제와 노이즈를 빠르게 파악하는 출발점입니다. 다만 조사·어미·불용어가 상위에 나타날 수 있으므로, 결과를 그대로 의미 있는 키워드라고 단정하면 안 됩니다.

### 체크포인트 3 — 길이 탐색

가장 짧은 리뷰와 가장 긴 리뷰를 찾아보세요. 단순히 길다는 사실만으로 오류라고 판단할 수 없는 이유와, 실제 데이터에서 추가 확인할 항목을 한 가지씩 작성하세요.


### 체크포인트 3 답변

1. **가장 짧은 리뷰와 긴 리뷰**:
    - 가장 짧은 리뷰: `스토리는 평범하지만 음악이 좋네요.` (19자)
    - 가장 긴 리뷰: `정말 재미있어요! 배우들의 연기가 인상적이었습니다.` (28자)

2. **길이만으로 오류를 판단할 수 없는 이유**: 리뷰의 길이는 사용자의 주관적인 표현 방식에 따라 달라지며, 짧더라도 핵심적인 감성 정보(예: '최고!', '별로')를 포함할 수 있고, 길더라도 풍부한 묘사를 담고 있을 수 있기 때문입니다.

3. **추가 확인 항목**:
    - **중복 여부**: 시스템적으로 생성된 자동 문구나 광고성 글이 반복적으로 나타나는지 확인해야 합니다.
    - **특수문자 비율**: 의미 없는 자음/모음 반복(ㅋㅋㅋ, ㅠㅠ)이나 특수기호가 비정상적으로 많은지 확인이 필요합니다.

```python
def simple_tokenize(text):
    """한글/영문/숫자 덩어리를 추출하는 매우 단순한 교육용 토크나이저"""
    return re.findall(r'[가-힣A-Za-z0-9]+', text.lower())

# TODO 7. 모든 리뷰에서 토큰을 모아 all_tokens 리스트를 만드세요.
all_tokens = [token for text in df['text'] for token in simple_tokenize(text)]

freq = Counter(all_tokens)
print('전체 토큰 수:', len(all_tokens))
print('고유 토큰 수:', len(freq))
print('어휘 다양성(type-token ratio):', round(len(freq) / len(all_tokens), 3))
display(pd.DataFrame(freq.most_common(15), columns=['token', 'count']))
```

```python
# 상위 토큰 시각화
# TODO 8. freq의 상위 12개 토큰으로 token, count 열을 가진 데이터프레임을 만드세요.
freq_df = pd.DataFrame(freq.most_common(12), columns=['token', 'count'])
freq_df = freq_df.sort_values('count')

plt.figure(figsize=(9, 5))
plt.barh(freq_df['token'], freq_df['count'], color='#6C5CE7')
plt.title('상위 토큰 빈도 (단순 토큰화)')
plt.xlabel('빈도')
plt.show()
```

### 체크포인트 4 — 빈도와 어휘 다양성

상위 빈도 토큰 중 감성 판단에 직접 도움이 되지 않을 수 있는 토큰을 하나 이상 고르세요. 이를 처리할 때의 장점과 단점을 각각 한 가지씩 적으세요.


### 체크포인트 4 답변

1. **감성 판단에 도움이 되지 않는 토큰**: `보고`, `영화`, `정말` 등
    - `보고`는 '영화 내용을 보고' 또는 '다시는 보고 싶지 않다' 등 긍정/부정 문맥 모두에서 자주 등장할 수 있어 단독으로는 변별력이 낮습니다.
    - `정말`은 강조어로서 긍정('정말 좋다')과 부정('정말 별로다') 모두에 쓰일 수 있습니다.

2. **처리 시 장점 (불용어 제거 시)**: 데이터의 노이즈를 줄여 모델이 '재미있어요', '지루했고'와 같이 감성이 뚜렷한 실질적인 키워드에 더 집중하게 함으로써 분석의 정확도를 높일 수 있습니다.

3. **처리 시 단점**: '보고 싶다'와 '보고 싶지 않다'처럼 강조어나 보조 용언이 전체 문맥의 감도를 결정하는 중요한 단서가 될 수 있는데, 이를 제거하면 미세한 감정의 차이나 강조의 정도를 놓칠 위험이 있습니다.

## 5. 한글 텍스트와 인코딩 확인

컴퓨터는 문자를 코드로 저장합니다. UTF-8은 가장 널리 쓰이는 문자 인코딩 중 하나이며, 한글을 포함한 다양한 문자를 표현할 수 있습니다. 파일을 잘못된 인코딩으로 읽으면 `�`처럼 깨진 문자가 나타날 수 있습니다.

또한 겉보기에는 같은 글자라도 유니코드 정규화 방식에 따라 내부 코드가 다를 수 있습니다. 데이터 병합·중복 제거 전에 확인할 가치가 있습니다.

```python
import unicodedata

text = '한글 NLP'
print('원문:', text)
utf8_bytes = text.encode('utf-8')
print('UTF-8 바이트:', utf8_bytes)

# TODO 9. UTF-8 바이트를 다시 문자열로 복원하세요.
restored = utf8_bytes.decode('utf-8')
print('UTF-8로 복원:', restored)

# 유니코드 정규화 예시
composed = '가'
decomposed = unicodedata.normalize('NFD', composed)
print('겉보기 문자열:', composed, decomposed)
print('문자열 동일 여부:', composed == decomposed)

# TODO 10. 두 문자열을 NFC로 정규화한 뒤 동일한지 확인하세요.
print('NFC 정규화 후 동일 여부:', unicodedata.normalize('NFC', composed) == unicodedata.normalize('NFC', decomposed))
```

```python
# 간단한 품질 점검 함수: 빈 텍스트, 깨진 문자, URL/이메일/전화번호처럼 보이는 패턴을 표시
def quality_flags(text):
    flags = []
    if not isinstance(text, str) or not text.strip():
        flags.append('빈 텍스트')
    if '' in text:
        flags.append('인코딩 깨짐 의심')
    if re.search(r'https?://\S+|www\.\S+', text):
        flags.append('URL 포함')
    if re.search(r'[\w.+-]+@[\w-]+\.[\w.-]+', text):
        flags.append('이메일 포함 가능성')
    # TODO 11. 전화번호로 보이는 패턴을 탐지하는 정규표현식을 완성하세요.
    if re.search(r'\d{2,3}[- ]?\d{3,4}[- ]?\d{4}', text):
        flags.append('전화번호 포함 가능성')
    return ', '.join(flags) if flags else '이상 없음'

df['quality_check'] = df['text'].apply(quality_flags)
display(df[['review_id', 'text', 'quality_check']])
```

## 6. 미니 실습: 감성별 자주 등장하는 토큰 비교

아래 분석은 **매우 단순한 탐색**입니다. 리뷰 수가 적고, 형태소 분석·불용어 처리·부정 표현 처리도 하지 않았습니다. 따라서 인과관계나 일반적 성능을 주장할 수 없습니다.

### 체크포인트 5 — 인코딩·품질·윤리

UTF-8 인코딩 또는 NFC 정규화를 통일하지 않으면 중복 제거·검색·모델 입력에서 어떤 문제가 생길 수 있나요? 또한 URL·이메일·전화번호를 탐지하는 이유를 개인정보 관점에서 설명하세요.


### 체크포인트 5 답변

1. **인코딩 및 정규화 미통일 시 문제점**:
    - **중복 제거 실패**: '가'(NFC)와 'ㄱ+ㅏ'(NFD)는 겉보기엔 같으나 바이트 코드가 달라 중복 데이터로 인식되지 않습니다.
    - **검색 및 모델 입력 오류**: 사용자가 입력한 검색어와 데이터셋의 인코딩이 다르면 검색 결과가 나오지 않으며, 모델 학습 시 동일한 단어를 서로 다른 토큰으로 임베딩하여 모델의 효율성과 성능이 저하됩니다.

2. **개인정보 탐지 이유**:
    - URL, 이메일, 전화번호 등은 특정 개인을 식별할 수 있는 개인정보(PII)에 해당합니다. 모델 학습 과정에서 이러한 정보가 포함될 경우, 모델이 개인정보를 암기하여 생성 결과로 노출하는 보안 사고가 발생할 수 있으므로 사전에 마스킹하거나 제거하여 프라이버시를 보호해야 합니다.

```python
def top_tokens_by_label(data, label, n=10):
    texts = data.loc[data['label'] == label, 'text']
    tokens = [tok for text in texts for tok in simple_tokenize(text)]
    return pd.DataFrame(Counter(tokens).most_common(n), columns=['token', 'count'])

# TODO 12. 긍정(1)과 부정(0) 리뷰의 상위 토큰을 각각 구하세요.
positive_top = top_tokens_by_label(df, 1)
negative_top = top_tokens_by_label(df, 0)

print('긍정 리뷰 상위 토큰')
display(positive_top)
print('부정 리뷰 상위 토큰')
display(negative_top)
```

```python
# @title 2. quality_flags에 반복 문자('ㅋㅋㅋㅋ', '!!!!', '...' 등) 탐지 규칙 추가
def quality_flags(text):
    flags = []

    if not isinstance(text, str) or not text.strip():
        flags.append('빈 텍스트')
        return ', '.join(flags)

    # 깨진 유니코드 문자의 대표적인 표시
    if '\ufffd' in text:
        flags.append('인코딩 깨짐 의심')

    if re.search(r'https?://\S+|www\.\S+', text):
        flags.append('URL 포함')

    if re.search(r'[\w.+-]+@[\w-]+\.[\w.-]+', text):
        flags.append('이메일 포함 가능성')

    if re.search(r'\d{2,3}[- ]?\d{3,4}[- ]?\d{4}', text):
        flags.append('전화번호 포함 가능성')

    # 같은 문자가 3회 이상 연속으로 반복되는 경우 탐지
    # 예: ㅋㅋㅋㅋ, !!!!, ..., ㅠㅠㅠ
    if re.search(r'(.)\1{2,}', text):
        flags.append('반복 문자 포함')

    return ', '.join(flags) if flags else '이상 없음'


df['quality_check'] = df['text'].apply(quality_flags)

print('=== 전체 리뷰 품질 점검 ===')
display(df[['id', 'text', 'quality_check']])

print('\n=== 반복 문자 규칙 테스트 ===')
test_texts = [
    '영화가 너무 웃겨요ㅋㅋㅋㅋ',
    '정말 최고!!!!',
    '결말이 조금 아쉽네요...',
    '일반적인 문장입니다.'
]
test_result = pd.DataFrame({
    'text': test_texts,
    'quality_check': [quality_flags(text) for text in test_texts]
})
display(test_result)

print('''
[오탐 가능 사례]
'...'은 문장 끝의 말줄임표처럼 정상적인 문장부호 표현일 수 있지만, 현재 규칙은 같은 문자가 3회 이상 반복되면 모두 '반복 문자 포함'으로 표시합니다.
또한 'ㅋㅋㅋ', 'ㅠㅠㅠ'처럼 감정을 전달하는 의미 있는 표현도 노이즈로 오탐할 수 있으므로, 실제 전처리에서는 탐지 후 무조건 삭제하기보다 별도로 보존하거나 반복 횟수만 정규화하는 방법을 고려할 수 있습니다.
''')
```
